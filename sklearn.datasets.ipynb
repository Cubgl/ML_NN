#### datasets.make_circles

```python
circles = datasets.make_circles()
```

```python
type(circles[0])
```

```python
print("features:", circles[0][:10])
print("target:", circles[1][:10])
```

```python
plt.figure(figsize=(4, 4))
plt.scatter(circles[0][:,0], circles[0][:, 1], c = circles[1])
```

```python
def plot_2d_dataset(data, target):
    plt.figure(figsize=(8, 8))
    plt.scatter(data[:, 0], data[:, 1], c = target)
```

```python
noisy_circles = datasets.make_circles(noise = 0.05)
```

```python
noisy_circles
```

```python
plot_2d_dataset(noisy_circles[0], noisy_circles[1])
```

#### datasets.make_classification

```python
simple_classification_problem = datasets.make_classification(n_features = 2, n_informative = 1, 
                                                            n_redundant = 1, n_clusters_per_class = 1,
                                                            random_state = 42 )
```

```python
simple_classification_problem
```

```python
plot_2d_dataset(simple_classification_problem[0], simple_classification_problem[1])
```

```python
classification_problem = datasets.make_classification(n_samples=300, n_features = 2, n_informative = 2, n_classes = 3, 
                                                      n_redundant = 0, n_clusters_per_class = 1,
                                                      random_state = 21)

```

```python
classification_problem
```

```python
plot_2d_dataset(classification_problem[0], classification_problem[1])
```

### Визуализация выбокри

```python
from pandas import DataFrame
```

```python
iris_frame = DataFrame(iris.data)
iris_frame.columns = iris.feature_names
iris_frame['target'] = iris.target
```

```python
iris_frame.head()
```

```python
iris_frame.target = iris_frame.target.apply(lambda x : iris.target_names[x])
```

```python
iris_frame.head()
```

```python
iris_frame[iris_frame.target == 'setosa'].hist('sepal length (cm)')
```

```python
plt.figure(figsize=(20, 24))
plot_number = 0
for feature_name in iris['feature_names']:
    for target_name in iris['target_names']:
        plot_number += 1
        plt.subplot(4, 3, plot_number)
        plt.hist(iris_frame[iris_frame.target == target_name][feature_name])
        plt.title(target_name)
        plt.xlabel('cm')
        plt.ylabel(feature_name[:-4])
```
